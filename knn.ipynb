```python
import math
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from collections import Counter

import warnings
warnings.filterwarnings('ignore')
```

```python
class KNN:
  def __init__(self,n_neighbours=3,task='regression'):
    self.n_neighbours=n_neighbours
    self.task=task

  def fit(self,X,y):
    self.X=X
    self.y=y

  def predict(self,X):
    y_pred=[]
    for x in X:
      y_pred.append(self._predict_single(x))
    return y_pred

  def _predict_single(self,x):
    dist=np.sqrt(np.sum((self.X-x)**2,axis=1))
    idx=np.argsort(dist)[:self.n_neighbours]
    if self.task=='regression':
      return np.mean(self.y[idx])
    else:
      most_common=Counter(self.y[idx]).most_common(1)
      return most_common[0][0]
```

```python
X=np.array([[1],[2],[3],[4],[5]])
y=np.array([1,0,1,0,1])

model=KNN(n_neighbours=2, task='classification')
model.fit(X, y)

X_test=np.array([[1.5], [3.5]])
preds=model.predict(X_test)
print("Regression predictions:", preds)
```
